```python
# Employee Training Tracker
# Domain: HR
# Console-based Python application using OOPS, data structures, and file handling

import json
import os
from datetime import datetime

DATA_FILE = "training_data.json"

# -----------------------------
# Data Models (OOPS)
# -----------------------------
class Employee:
    def __init__(self, emp_id, name, department):
        self.emp_id = emp_id
        self.name = name
        self.department = department

    def to_dict(self):
        return {
            "emp_id": self.emp_id,
            "name": self.name,
            "department": self.department
        }


class Training:
    def __init__(self, training_id, title, date):
        self.training_id = training_id
        self.title = title
        self.date = date
        self.attendance = []  # list of emp_ids

    def mark_attendance(self, emp_id):
        if emp_id not in self.attendance:
            self.attendance.append(emp_id)

    def to_dict(self):
        return {
            "training_id": self.training_id,
            "title": self.title,
            "date": self.date,
            "attendance": self.attendance
        }


# -----------------------------
# Data Storage Handler
# -----------------------------
class FileManager:
    @staticmethod
    def load_data():
        if not os.path.exists(DATA_FILE):
            return {"employees": {}, "trainings": {}}
        with open(DATA_FILE, "r") as f:
            return json.load(f)

    @staticmethod
    def save_data(data):
        with open(DATA_FILE, "w") as f:
            json.dump(data, f, indent=4)


# -----------------------------
# Main Application Logic
# -----------------------------
class TrainingTracker:
    def __init__(self):
        self.data = FileManager.load_data()

    # ---- Employee CRUD ----
    def add_employee(self):
        emp_id = input("Enter Employee ID: ")
        name = input("Enter Employee Name: ")
        dept = input("Enter Department: ")

        emp = Employee(emp_id, name, dept)
        self.data["employees"][emp_id] = emp.to_dict()
        FileManager.save_data(self.data)
        print("Employee added successfully!\n")

    def view_employees(self):
        print("\n--- Employee List ---")
        for emp in self.data["employees"].values():
            print(emp)
        print()

    # ---- Training CRUD ----
    def add_training(self):
        training_id = input("Enter Training ID: ")
        title = input("Enter Training Title: ")
        date = input("Enter Date (YYYY-MM-DD): ")

        training = Training(training_id, title, date)
        self.data["trainings"][training_id] = training.to_dict()
        FileManager.save_data(self.data)
        print("Training registered successfully!\n")

    def view_trainings(self):
        print("\n--- Training List ---")
        for tr in self.data["trainings"].values():
            print(tr)
        print()

    # ---- Attendance ----
    def mark_attendance(self):
        training_id = input("Enter Training ID: ")
        emp_id = input("Enter Employee ID: ")

        if training_id in self.data["trainings"] and emp_id in self.data["employees"]:
            self.data["trainings"][training_id]["attendance"].append(emp_id)
            FileManager.save_data(self.data)
            print("Attendance marked successfully!\n")
        else:
            print("Invalid Training ID or Employee ID\n")

    # ---- Reports ----
    def generate_report(self):
        print("\n--- Training Report ---")
        for tr in self.data["trainings"].values():
            print(f"Training: {tr['title']} ({tr['date']})")
            print("Attendees:")
            for emp_id in tr["attendance"]:
                emp = self.data["employees"].get(emp_id)
                if emp:
                    print(f" - {emp['name']} ({emp['department']})")
            print()

    # ---- Menu ----
    def menu(self):
        while True:
            print("""
Employee Training Tracker
1. Add Employee
2. View Employees
3. Register Training
4. View Trainings
5. Mark Attendance
6. Generate Report
7. Exit
""")
            choice = input("Enter choice: ")

            if choice == "1":
                self.add_employee()
            elif choice == "2":
                self.view_employees()
            elif choice == "3":
                self.add_training()
            elif choice == "4":
                self.view_trainings()
            elif choice == "5":
                self.mark_attendance()
            elif choice == "6":
                self.generate_report()
            elif choice == "7":
                print("Exiting system...")
                break
            else:
                print("Invalid choice!\n")


# -----------------------------
# Program Entry Point
# -----------------------------
if __name__ == "__main__":
    app = TrainingTracker()
    app.menu()
```
